# Tensor  

### Zero-Dimensional Tensor (0D)

```python
import numpy as np
```

```python
scalar = np.array(6)
```

```python
print(scalar)
print(scalar.ndim)
print(scalar.shape)
print(scalar.size)
```

ndim = 0 means there are zero axes.
shape = () means there are no axes with a length.
size = 1 means it contains one element.

### One-Dimensional Tensor (1D)

```python
vector = np.array([1, 2, 3, 4])
```

```python
print(vector.ndim)
print(vector.shape)
print(vector.size)
```

### Two-Dimensional Tensor (2D)

```python
matrix = np.array([[1, 2, 3],[4, 5, 6]])
```

```python
print(matrix.ndim)
print(matrix.shape)
print(matrix.size)
```

### Three-Dimensional Tensors (3D)

```python
tensor_3d = np.array([[[1, 2], [3, 4]],   [[5, 6], [7, 8]]])
```

```python
print(tensor_3d.ndim)
print(tensor_3d.shape)
print(tensor_3d.size)
```
